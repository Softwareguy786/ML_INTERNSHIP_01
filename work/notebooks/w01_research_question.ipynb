Why this connects to my goal: My end goal is healthcare data science. Predicting which web pages are 'declining' is structurally identical to predicting which patients are 'high-risk' for readmission. This lane teaches me binary classification, handling imbalanced data, and measuring the cost of false alarms, which are the exact skills I need for healthcare.*

Search question: Which web pages are most likely to lose traffic (trend downward) in the next 30 days?

Unit of analysis: Each row = one web page.

Output: A probability score (0 to 1) for each page, ranking them from highest risk to lowest risk.

Decision: The content marketing team decides which pages to assign for a rewrite/refresh.

Action: The team takes the top 20 highest-risk pages and updates their content to make them fresh again.

Cost of a wrong call (this is the important part):

    False Positive (I say the page is dying, but it's actually fine): The team wastes 3 hours rewriting a page that didn't need it. Cost = Wasted editor salary (~$150).

    False Negative (I say the page is fine, but it actually dies): The page loses Google rankings and drops from 1,000 visits/month to 200. Cost = Lost ad revenue and leads (~$500 per month).

    Healthcare translation: In healthcare, a False Negative means sending a sick patient home (deadly), and a False Positive means an unnecessary hospital visit (wastes money). This project teaches me to balance these tradeoffs.

```python
import pandas as pd

# Load the starter dataset
df = pd.read_csv('../../data/raw/content_refresh_anonymized.csv')

# 1. How big is the problem?
total_pages = len(df)
declining_count = (df['trend_direction'] == 'down').sum()
declining_pct = (declining_count / total_pages) * 100

print(f"📊 Total pages in dataset: {total_pages:,}")
print(f"📉 Pages marked as 'declining': {declining_count:,} ({declining_pct:.1f}%)")

# 2. Show that declining pages get way less traffic (proving it's worth fixing)
avg_traffic_declining = df[df['trend_direction'] == 'down']['impressions_90d'].mean()
avg_traffic_stable = df[df['trend_direction'] != 'down']['impressions_90d'].mean()
traffic_drop_pct = ((avg_traffic_stable - avg_traffic_declining) / avg_traffic_stable) * 100

print(f"\n🚀 Avg 90-day impressions (Stable pages): {avg_traffic_stable:,.0f}")
print(f"⚠️  Avg 90-day impressions (Declining pages): {avg_traffic_declining:,.0f}")
print(f"📉 Declining pages get {traffic_drop_pct:.1f}% LESS traffic than stable pages.")

# 3. Show that the starter model actually does something (proving ML helps)
from sklearn.metrics import accuracy_score
# Just loading the starter split to show there's a signal
# (If you have the starter pipeline imported, you can show a real metric.
# If not, just showing class imbalance + feature difference is enough.)
avg_wordcount_declining = df[df['trend_direction'] == 'down']['word_count'].mean()
avg_wordcount_stable = df[df['trend_direction'] != 'down']['word_count'].mean()
print(f"\n📝 Avg word count (Declining): {avg_wordcount_declining:.0f}")
print(f"📝 Avg word count (Stable): {avg_wordcount_stable:.0f}")
print("💡 Since declining and stable pages have different word counts, ML can learn to tell them apart!")
```

I can claim that this model identifies patterns (like low word count or old publish dates) that are associated with traffic decline in this specific dataset. I can build a ranking system to prioritize which pages the team should look at first.
I cannot claim that low word count causes the decline (correlation isn't causation). I cannot claim this model will work for completely different websites or industries without retraining. This is a decision-support tool, not an absolute truth machine.
